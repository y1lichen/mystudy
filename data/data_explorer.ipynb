```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
csv_path = "ChillerPlant.csv"
df = pd.read_csv(csv_path)
df = df.fillna(0.0)
```

```python
df.columns
```

```python
# random_state=42 確保每次切出來的結果一樣，方便重現實驗
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

print(f"總資料筆數: {len(df)}")
print(f"訓練集 (Train): {len(train_df)} 筆")
print(f"測試集 (Test): {len(test_df)} 筆")

# 存成獨立的 CSV 檔案
train_df.to_csv("ChillerPlant_train.csv", index=False)
test_df.to_csv("ChillerPlant_test.csv", index=False)
print("成功儲存隨機切分的 Train 和 Test CSV！")
```

```python
# 1. 找出主機 2 啟動的索引 (0 變 1)
df['CH2_Start'] = (df['CHL_STA_2'] == 1) & (df['CHL_STA_2'].shift(1) == 0)
ch2_start_indices = df[df['CH2_Start'] == True].index

print(f"主機 2 總共啟動了 {len(ch2_start_indices)} 次")

# 2. 分析第一次啟動時的電力浪湧 (Power Surge)
if len(ch2_start_indices) > 0:
    idx = ch2_start_indices[0]
    # 觀察啟動前 2 分鐘到啟動後 10 分鐘的電力與溫度變化
    startup_profile = df.loc[idx-2 : idx+10, 
                             ['Datetime', 'CHL_STA_2', 'CHL_POW_2', 'CHL_SW_TEMP_2', 'CHL_RW_TEMP_2']]
    print("\n主機 2 第一次啟動剖面：")
    print(startup_profile)
```

```python
# 確保 Datetime 欄位為時間格式
df['Datetime'] = pd.to_datetime(df['Datetime'])

# 找出起始與結束時間
start_time = df['Datetime'].min()
end_time = df['Datetime'].max()

# 計算總時長
duration = end_time - start_time

print(f"數據集起始時間: {start_time}")
print(f"數據集結束時間: {end_time}")
print(f"總時間跨度: {duration}")

# 檢查採樣頻率（查看前兩筆的時間差）
sample_gap = df['Datetime'].iloc[1] - df['Datetime'].iloc[0]
print(f"資料採樣間隔: {sample_gap}")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 確保時間格式正確
df['Datetime'] = pd.to_datetime(df['Datetime'])

# 建立 3 個子圖，分別顯示主機、水塔、水泵的啟停
fig, axes = plt.subplots(3, 1, figsize=(20, 15), sharex=True)

# --- 子圖 1: 冷凍主機 (Chillers) ---
for i in range(1, 4):
    axes[0].plot(df['Datetime'], df[f'CHL_STA_{i}'], label=f'Chiller {i}', alpha=0.8)
axes[0].set_title('Annual Chiller Status (CHL_STA_1~3)')
axes[0].set_ylabel('Status (0=Off, 1=On)')
axes[0].legend(loc='upper right')

# --- 子圖 2: 冷卻水塔 (Cooling Towers) ---
for i in range(1, 4):
    axes[1].plot(df['Datetime'], df[f'CT_STA_{i}'], label=f'Tower {i}', alpha=0.8)
axes[1].set_title('Annual Cooling Tower Status (CT_STA_1~3)')
axes[1].set_ylabel('Status (0=Off, 1=On)')
axes[1].legend(loc='upper right')

# --- 子圖 3: 二次側水泵 (Secondary Pumps) ---
for i in range(1, 3):
    axes[2].plot(df['Datetime'], df[f'CWL_SEC_PM_STA_{i}'], label=f'Sec Pump {i}', alpha=0.8)
axes[2].set_title('Annual Secondary Pump Status (CWL_SEC_PM_STA_1~2)')
axes[2].set_ylabel('Status (0=Off, 1=On)')
axes[2].legend(loc='upper right')

plt.xlabel('Date')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# 載入資料 (請確保檔案在路徑中)
df = pd.read_csv("ChillerPlant.csv")

# 1. 自動識別所有水泵電力欄位
# 通常包含 'PM' (Pump) 或 'PUMP' 且包含 'POW' (Power)
pump_power_cols = [col for col in df.columns if 'POW' in col and ('PM' in col or 'PUMP' in col)]

print(f"偵測到的水泵電力欄位: {pump_power_cols}")

# 2. 計算各欄位為 0 的比例
zero_stats = []
for col in pump_power_cols:
    zero_count = (df[col] == 0).sum()
    total_count = len(df)
    proportion = (zero_count / total_count) * 100
    zero_stats.append({
        "欄位名稱": col,
        "零值筆數": zero_count,
        "總筆數": total_count,
        "零值比例 (%)": round(proportion, 2)
    })

# 3. 轉換為 DataFrame 顯示結果
zero_stats_df = pd.DataFrame(zero_stats)
print("\n水泵功率為 0 的比例分析：")
print(zero_stats_df)

# 如果你想看「所有水泵同時全關」的比例
if pump_power_cols:
    all_off_proportion = (df[pump_power_cols].sum(axis=1) == 0).mean() * 100
    print(f"\n所有水泵同時關閉 (功率皆為 0) 的比例: {all_off_proportion:.2f}%")
```
